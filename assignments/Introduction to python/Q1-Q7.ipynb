```python
# Q1. Write a python program to sum of the first n positive integers.

I = int(input("enter your number:"))
sum = 0
for i in range (1,I+1):
    sum += i
    
print(sum)
```

```python
#Q2. write a Python program to count occurrences of a substring in a string

string = "my name is khan KHAN"
sub = "khan" 
count= 0

string = string.split()

for i in string:
    if i.lower() == sub:
        count+=1 

print(count)
```

```python
#Q=3 Write a Python program to count the occurrences of each word in a givensentence.

string = input("enter sentence:")
w = {}


string= string.split()
for i in string:
    if i in w.keys():
        w[i]+= 1
    else:
        w[i]= 1

print(w)

        



```

```python
#Q5 Write a Python program to add 'ing' at the end of a given string (length should be at least 3). If the given string already ends with 'ing' then add 'ly' instead If the string length of the given string is less than 3, leave it unchanged

St = input("enter string : ")

if len(St) < 3:
    print(St)
elif St.endswith('ing'):
    print(St + 'ly')
else:
    print(St + 'ing')
```
