```python
#Q15. Given a number n, write a python program to make and print the list of Fibonacci series up to n. Input : n=7 Hint : first 7 numbers in the series Expected output :First few Fibonacci numbers are 0, 1, l, 2, 3, 5, 8, 13

n = int(input("enter any number n: "))
a,b = 0 , 1
fib = []

for i in range(n+1):
    fib.append(a)
    a, b = b, a+b
print(fib)
```

```python
#Q16. Counting the frequencies in a list using a dictionary in Python. Input : [1, 1, 1, 5, 5, 3, 1, 3, 3, 1,4, 4, 4, 2, 2, 2, 2] Expected output : 1 : 5 , 2 : 4 , 3 : 3 , 4 : 3 , 5 : 2

lst = [1, 1, 1, 5, 5, 3, 1, 3, 3, 1, 4, 4, 4, 2, 2, 2, 2]
di = {}

for i in lst:
    di[i] = di.get(i, 0) + 1
print(di)
```

```python
#Q17. Write a python program using function to find the sum of odd series and even series Odd series: 12/ l! +32/ 3! + 52/5!.....n Even series: 22/ 2! + 42/ 4! + 62/6!....n
import math

def odd(n):
    total = 0
    for i in range(1, n+1, 2):
        total += (i**2) / math.factorial(i)
    return total

def even(n):
    total = 0
    for i in range(2, n+1, 2):
        total += (i**2) / math.factorial(i)
    return total

print("Sum of odd series:", odd(n))
print("Sum of even series:", even(n))
```

```python
#Q18. Python Program to Find Factorial of Number Using Recursion

def recursion(num):
    if num == 0 or num == 1:
        return 1
    else:
        return num * recursion(num - 1)

print(f"Factorial of {n} is {recursion(n)}")
```

```python
#Q19. Write a Python function that takes a list and returns a new list with unique elements of the first list.

li = [1,2,2,4,5,67,68,34,35,35]
def unique(lis):
    return list(set(lis))

print(unique(li))
```
