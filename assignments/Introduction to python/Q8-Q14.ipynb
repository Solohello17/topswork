```python
#Q8. Write a Python program to check whether a list contains a sublist.

l1 = [23,45,99,100,101,44,58]

l2 = [45,58]

count = 0
for i in l2:
    if i in l1:
        count +=1

if count == len(l2):
    print(True)
else:
    print(False)



```

```python
# Q10 Write a Python program to get unique values from a list.

l4= [10,20,20,"java","python",45]

l5 = []

for i in l4:
    if l4.count(i) == 1:
        l5.append(i)

print(l5)

```

```python
# 11. Write a Python program to unzip a list of tuples into individual lists.

l6 = [(1, 'a', True), (2, 'b', False), (3, 'c', True)]

a,b,c = list(zip(*l6))

# Convert each tuple to a list
print(list(a))
print(list(b))
print(list(c))
```

```python
#Q.13 Write a Python program to sort a dictionary (ascending descending) by value.

di = {'a':1,'b':2 ,'d':4, 'f':6, 'c':3,'e':5}
list = []

for k,v in di.items():
    list.append((k,v))
print(list)
list.sort()
sorted_di = {}

for k,v in list:
    sorted_di[k]=v

print(sorted_di)

list.sort(reverse=True)
desc_di = {}
for k,v in list:
    desc_di[k]=v

print(desc_di)
```

```python
#Q14. . Write a Python program to find the highest 3 values in a dictionary.

data = {'a':1,'b':2 ,'d':4, 'f':6, 'c':3,'e':5}
lis = []
for k,v in data.items():
    lis.append((k,v))

lis.sort(reverse=True)
top3= lis[:3]
sorted_data = {}

for k,v in top3:
    sorted_data[k]=v

print(sorted_data)
```
